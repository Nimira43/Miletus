# Simulate an AR (Auto Regression) process

simulate-an-ar-process.png

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
alpha = np.array([-0.6, 0.9])
k = len(alpha)
N = 200
x = np.zeros(N)

for i in range(k, N):
  x[i] = sum(alpha * x[i-k:i]) + np.random.randn()

plt.plot(x)
plt.xlabel('X axis - time')
plt.title('AR(%s) process' %k)
plt.show()
```
